```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(img, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    imagen_c=cv2.filter2D(img,-1,kernel)
    return imagen_c

def convolucion_media_v(img, tkernel):
    kernel = np.ones((tkernel, 1), np.float32) / tkernel
    imagen_c=cv2.filter2D(img,-1,kernel)
    return imagen_c

def convolucion_media_d(img, tkernel):
    kernel = np.eye(tkernel, dtype=np.float32) / tkernel
    imagen_c=cv2.filter2D(img,-1,kernel)
    return imagen_c

def convolucion_media_di(img, tkernel):
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32)) / tkernel
    imagen_c=cv2.filter2D(img,-1,kernel)
    return imagen_c

img = cv2.imread('images.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_com=convolucion_media_h(img_rgb,29)
img_com1=convolucion_media_v(img_rgb,29)
img_com2=convolucion_media_d(img_rgb,29)
img_com3=convolucion_media_di(img_rgb,29)

plt.figure(figsize=(10,10))
plt.subplot(321)
plt.title("Original")
plt.imshow(img_rgb)
plt.subplot(322)
plt.title("Horizontal")
plt.imshow(img_com)
plt.subplot(323)
plt.title("Vertical")
plt.imshow(img_com1)
plt.subplot(324)
plt.title("Diagonal")
plt.imshow(img_com2)
plt.subplot(325)
plt.title("Diagonal Inversa")
plt.imshow(img_com3)
plt.subplot(326)
plt.title("Original de Nuevo")
plt.imshow(img_rgb)

plt.show()


```
