```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread('images.jpg',0)
kernel=np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])#mascara de convoluciones
img_fin=cv2.filter2D(img,-1,kernel)
plt.subplot(121)
plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread('images.jpg')
kernel=np.ones((5,5),np.float32)/25
suavizado=cv2.filter2D(img,-1,kernel)
kernel2=np.ones((7,7),np.float32)/49
suavizado2=cv2.filter2D(img,-1,kernel2)
plt.subplot(131)
plt.imshow(img,cmap='gray')
plt.subplot(132)
plt.imshow(suavizado,cmap='gray')
plt.subplot(133)
plt.imshow(suavizado2,cmap='gray')
plt.show()
```
