```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(img, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    return cv2.filter2D(img, -1, kernel)

def convolucion_media_v(img, tkernel):
    kernel = np.ones((tkernel, 1), np.float32) / tkernel
    return cv2.filter2D(img, -1, kernel)

img = cv2.imread('moire2.png', 0)
img_suave = cv2.GaussianBlur(img, (0, 0), 1.5)
img_h = convolucion_media_h(img_suave, 10)
img_hv = convolucion_media_v(img_h, 10)

plt.figure(figsize=(12,6))
plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Imagen original")
plt.axis("off")
plt.subplot(122)
plt.imshow(img_hv, cmap='gray')
plt.title("Vertical + Horizontal")
plt.axis("off")
plt.show() 
```

Aliasing

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('al2.jpeg', 0)

img_gauss = cv2.GaussianBlur(img, (7, 7), 0)
img_media = cv2.blur(img, (5, 5))
img_bilateral = cv2.bilateralFilter(img, 9, 75, 75)

plt.figure(figsize=(16, 10))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_gauss, cmap='gray')
plt.title('Filtro Gaussiano')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_media, cmap='gray')
plt.title('Filtro de Media')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img_bilateral, cmap='gray')
plt.title('Filtro Bilateral')
plt.axis('off')

plt.tight_layout()
plt.show()
```
