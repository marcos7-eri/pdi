```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [6, 6]

alto, ancho = 100, 100
imagen = np.zeros((alto, ancho, 3), dtype=np.uint8)

mitad_alto = alto // 2
mitad_ancho = ancho // 2

for i in range(alto):
    for j in range(ancho):
        if i < mitad_alto and j < mitad_ancho:
            imagen[i, j] = [255, 0, 0]
        elif i < mitad_alto and j >= mitad_ancho:
            imagen[i, j] = [0, 255, 0]
        elif i >= mitad_alto and j < mitad_ancho:
            imagen[i, j] = [0, 0, 255]
        else:
            imagen[i, j] = [255, 255, 0]

plt.imshow(imagen)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [6, 6]

bloque = 50
size = bloque * 3

imagen = np.zeros((size, size, 3), dtype=np.uint8)

colores_base = [
    [183, 18, 52],
    [0, 155, 72],
    [255, 255, 255],
    [255, 88, 0],
    [255, 213, 0],
    [0, 70, 173]
]

colores = colores_base * 2
colores = colores[:9]

np.random.shuffle(colores)

k = 0

for i in range(3):
    for j in range(3):
        imagen[
            i*bloque:(i+1)*bloque,
            j*bloque:(j+1)*bloque
        ] = colores[k]
        k += 1

plt.imshow(imagen)
```
