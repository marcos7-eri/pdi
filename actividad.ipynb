```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('img1.jpeg')
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img1_gray = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
cv2.imwrite("img1_bgr_guardada.jpg", img1)
cv2.imwrite("img1_gris_guardada.jpg", img1_gray)

fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img1_rgb)
axs[0].set_title('Imagen Original RGB')
axs[0].axis("off")

red = np.zeros_like(img1_rgb)
red[...,0] = img1_rgb[...,0]
axs[1].imshow(red)
axs[1].set_title('Canal Rojo')
axs[1].axis("off")

green = np.zeros_like(img1_rgb)
green[...,1] = img1_rgb[...,1]
axs[2].imshow(green)
axs[2].set_title('Canal Verde')
axs[2].axis("off")

blue = np.zeros_like(img1_rgb)
blue[...,2] = img1_rgb[...,2]
axs[3].imshow(blue)
axs[3].set_title('Canal Azul')
axs[3].axis("off")

plt.show()

fig, axs = plt.subplots(1, 3, figsize=(18,5))

axs[0].imshow(img1_gray, cmap='gray')
axs[0].set_title('Escala de Grises')
axs[0].axis("off")

fig_img = axs[1].imshow(img1_gray, cmap='jet')
fig.colorbar(fig_img, ax=axs[1], shrink=0.5)
axs[1].set_title('Color Falso')

axs[2].imshow(img1_rgb)
axs[2].set_title('RGB')
axs[2].axis("off")

plt.show()
```
