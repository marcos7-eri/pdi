```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("saturada.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

h, w, _ = img_rgb.shape
bayer = np.zeros((h, w), dtype=np.uint8)

bayer[0::2,0::2] = img_rgb[0::2,0::2,1]
bayer[0::2,1::2] = img_rgb[0::2,1::2,2]
bayer[1::2,0::2] = img_rgb[1::2,0::2,0]
bayer[1::2,1::2] = img_rgb[1::2,1::2,1]

recon_bg = cv2.cvtColor(bayer, cv2.COLOR_BAYER_BG2RGB)
recon_gb = cv2.cvtColor(bayer, cv2.COLOR_BAYER_GB2RGB)
recon_rg = cv2.cvtColor(bayer, cv2.COLOR_BAYER_RG2RGB)
recon_gr = cv2.cvtColor(bayer, cv2.COLOR_BAYER_GR2RGB)

plt.figure(figsize=(12,6))
plt.subplot(131); plt.imshow(img_rgb); plt.title("Imagen Original"); plt.axis("off")
plt.subplot(132); plt.imshow(bayer, cmap="gray"); plt.title("Mosaico"); plt.axis("off")
plt.subplot(133); plt.imshow(recon_bg); plt.title("Reconstruccion Bayer"); plt.axis("off")
plt.show()

plt.figure(figsize=(12,8))
plt.subplot(221); plt.imshow(recon_bg); plt.title("Bayer BG → RGB"); plt.axis("off")
plt.subplot(222); plt.imshow(recon_gb); plt.title("Bayer GB → RGB"); plt.axis("off")
plt.subplot(223); plt.imshow(recon_rg); plt.title("Bayer RG → RGB"); plt.axis("off")
plt.subplot(224); plt.imshow(recon_gr); plt.title("Bayer GR → RGB"); plt.axis("off")
plt.show()

img_yuv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2YUV)
img_rgb2 = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(12,6))
plt.subplot(131); plt.imshow(img_rgb); plt.title("Original RGB"); plt.axis("off")
plt.subplot(132); plt.imshow(img_yuv); plt.title("RGB → YUV"); plt.axis("off")
plt.subplot(133); plt.imshow(img_rgb2); plt.title("YUV → RGB"); plt.axis("off")
plt.show()
```
