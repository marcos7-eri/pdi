```python
import cv2
import matplotlib.pyplot as plt

img_bayer = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

rgb_bg = cv2.cvtColor(img_bayer, cv2.COLOR_BAYER_BG2RGB)
rgb_gb = cv2.cvtColor(img_bayer, cv2.COLOR_BAYER_GB2RGB)
rgb_rg = cv2.cvtColor(img_bayer, cv2.COLOR_BAYER_RG2RGB)
rgb_gr = cv2.cvtColor(img_bayer, cv2.COLOR_BAYER_GR2RGB)

plt.figure(figsize=(12,8))
plt.subplot(221); plt.imshow(rgb_bg); plt.title("Bayer BG → RGB"); plt.axis("off")
plt.subplot(222); plt.imshow(rgb_gb); plt.title("Bayer GB → RGB"); plt.axis("off")
plt.subplot(223); plt.imshow(rgb_rg); plt.title("Bayer RG → RGB"); plt.axis("off")
plt.subplot(224); plt.imshow(rgb_gr); plt.title("Bayer GR → RGB"); plt.axis("off")
plt.show()

img_rgb = cv2.imread("umbralizacion.jpg")
img_rgb = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2RGB)

img_yuv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2YUV)
img_rgb2 = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(12,6))
plt.subplot(131); plt.imshow(img_rgb); plt.title("Original RGB"); plt.axis("off")
plt.subplot(132); plt.imshow(img_yuv); plt.title("YUV"); plt.axis("off")
plt.subplot(133); plt.imshow(img_rgb2); plt.title("Reconstruido RGB"); plt.axis("off")
plt.show()

img_yuv2 = cv2.cvtColor(img_rgb2, cv2.COLOR_RGB2YUV)

plt.figure(figsize=(8,4))
plt.subplot(121); plt.imshow(img_yuv); plt.title("YUV inicial"); plt.axis("off")
plt.subplot(122); plt.imshow(img_yuv2); plt.title("YUV reconstruido"); plt.axis("off")
plt.show()
```
