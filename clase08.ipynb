ECUALIZACION DE UN HISTOGRAMA "SE DESEA NORMALIZAR UN IMAGEN A UN PUNTO OPTIMO POR HISTOGRAMA"

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img=cv2.imread('colores.jpg',0)
histograma, calc=np.histogram(img.flatten(),256,[0,256])
distAcum=histograma.cumsum()
dis_norm=distAcum*histograma.max()/distAcum.max()
ecualizar=cv2.equalizeHist(img)
histograma_equ,_=np.histogram(ecualizar.flatten(),256,[0,256])

plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.imshow(img,cmap='gray')

plt.subplot(2,2,2)
plt.plot(distAcum,color='b')
#imprimir histogrma por plt
plt.hist(img.flatten(),256,[0,256],color='r')

plt.subplot(2,2,3)
plt.imshow(ecualizar,cmap='gray')

plt.subplot(2,2,4)
plt.plot(histograma_equ,color='b')

plt.show()
```
