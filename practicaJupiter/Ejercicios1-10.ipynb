```python
n = int(input("ingresa un numero: "))
fact = 1
for i in range(1, n + 1):
    fact = fact * i

print("el factorial es:", fact)
```

```python
n = int(input("ingresa un numero: "))
es_primo = True
if n <= 1:
    es_primo = False
else:
    for i in range(2, n):
        if n % i == 0:
            es_primo = False
            break

if es_primo:
    print("es primo")
else:
    print("no es primo")
```

```python
cant = int(input("cuantos numeros vas a ingresar (min 3): "))
nums = []
for i in range(cant):
    n = int(input("numero: "))
    nums.append(n)

mayor = max(nums)
menor = min(nums)

print("mayor:", mayor)
print("menor:", menor)
```

ej5

```python
nums = []
for i in range(7):
    n = int(input("numero: "))
    nums.append(n)

rep = {}
for n in nums:
    if n in rep:
        rep[n] += 1
    else:
        rep[n] = 1

max_rep = max(rep, key=rep.get)

print("numero que mas se repite:", max_rep)
print("veces:", rep[max_rep])
```

```python
nums = []
cant = int(input("cuanto numeros: "))

for i in range(cant):
    n = int(input("numero: "))
    nums.append(n)

for i in range(len(nums)):
    for j in range(len(nums) - 1):
        if nums[j] > nums[j + 1]:
            aux = nums[j]
            nums[j] = nums[j + 1]
            nums[j + 1] = aux

print("ordenados:", nums)
```
